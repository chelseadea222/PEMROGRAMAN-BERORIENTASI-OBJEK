```python
class Person:
  def __init__(self, nama, umur):
    self.nama = nama
    self.umur = umur

  def tampilkan_info(self):
   print(f"Nama: {self.nama}, Umur: {self.umur}")

class Mahasiswa(Person):
  def __init__(self, nama, umur, nim):
      super().__init__(nama, umur)
      self.nim = nim

  def tampilkan_mahasiswa(self):
    self.tampilkan_info()
    print(f"NIM: {self.nim}")

class Dosen(Person):
  def __init__(self, nama, umur, nidn):
    super().__init__(nama, umur)
    self.nidn = nidn

  def tampilkan_dosen(self):
      self.tampilkan_info()
      print(f"NIDN: {self.nidn}")

##Menampilkan hasil

mhs = Mahasiswa("Fadil", 20, "D12345")
mhs.tampilkan_mahasiswa()

dsn = Dosen("Dr.Budi", 40, "N67890")
dsn.tampilkan_dosen()
```

```python
class Barang:
  def __init__(self, id_barang, nama, harga):
    self.id_barang = id_barang
    self.nama = nama
    self.harga = harga
  def info(self):
    return f"[{self.id_barang}] {self.nama:15} | Rp{self.harga:10,}"

class BarangElektronik(Barang):
  def __init__(self, id_barang, nama, harga, garansi):
      super().__init__(id_barang, nama, harga)
      self.garansi = garansi

  def info(self):
    return f"{super().info()} | Garansi: {self.garansi} bln"

class BarangKonsumsi(Barang):
  def __init__(self, id_barang, nama, harga, tgl_exp):
    super().__init__(id_barang, nama, harga)
    self.tgl_exp = tgl_exp
  def info(self):
    return f"{super().info()} | Exp: {self.tgl_exp}"


laptop = BarangElektronik("E01", "Laptop Gaming", 15000000, 24)
susu = BarangKonsumsi("K05", "Susu UHT", 18000, "2026-12-01")

print(laptop.info())
print(susu.info())
```

```python
class Kendaraan:
  def __init__(self, id_kendaraan, merk, kecepatan):
    self.id_kendaraan = id_kendaraan
    self.merk = merk
    self.kecepatan = kecepatan

  def info(self):
    print(f"ID: {self.id_kendaraan}, Merk: {self.merk}, Kecepatan: {self.kecepatan}")

class Mobil(Kendaraan):
  def __init__(self, id_kendaraan, merk, kecepatan, jumlah_pintu):
      super().__init__(id_kendaraan, merk, kecepatan)
      self.jumlah_pintu = jumlah_pintu

  def info(self):
      super().info()
      print(f"Jumlah Pintu: {self.jumlah_pintu}")

class Motor(Kendaraan):
    def __init__(self, id_kendaraan, merk, kecepatan, jenis):
        super().__init__(id_kendaraan, merk, kecepatan)
        self.jenis = jenis

    def info(self):
        super().info()
        print(f"Jenis: {self.jenis}")

mobil1 = Mobil("M001", "Toyota", 120, 4)
motor1 = Motor("M002", "Honda", 100, "Sport")

print(mobil1.info())
print(motor1.info())
```

```python
# Parent Class
class Produk:
    def __init__(self, nama, harga, stok):
        self.nama = nama
        self.harga = harga
        self.stok = stok

    def info(self):
        print(f"Produk: {self.nama}")
        print(f"Harga: {self.harga}")
        print(f"Stok: {self.stok}")

    def hitung_total(self, jumlah):
        return self.harga * jumlah


# Child Class Elektronik
class Elektronik(Produk):
    def __init__(self, nama, harga, stok, garansi):
        super().__init__(nama, harga, stok)
        self.garansi = garansi

    def info(self):
        super().info()
        print(f"Garansi: {self.garansi} bulan")


# Child Class Pakaian
class Pakaian(Produk):
    def __init__(self, nama, harga, stok, ukuran):
        super().__init__(nama, harga, stok)
        self.ukuran = ukuran

    def info(self):
        super().info()
        print(f"Ukuran: {self.ukuran}")


# Dataset
produk_marketplace = [
    {"nama": "Laptop Gaming", "harga": 15000000, "kategori": "Elektronik", "garansi": 24, "stok": 5},
    {"nama": "Kemeja Flanel", "harga": 200000, "kategori": "Pakaian", "ukuran": "M", "stok": 10},
    {"nama": "Smart TV", "harga": 7000000, "kategori": "Elektronik", "garansi": 12, "stok": 3},
    {"nama": "Celana Jeans", "harga": 300000, "kategori": "Pakaian", "ukuran": "L", "stok": 7}
]

# Simulasi pembelian
laptop = Elektronik("Laptop Gaming", 15000000, 5, 24)
laptop.info()
jumlah_beli = 2
print(f"Total beli {jumlah_beli}: {laptop.hitung_total(jumlah_beli)}\n")

kemeja = Pakaian("Kemeja Flanel", 200000, 10, "M")
kemeja.info()
jumlah_beli = 3
print(f"Total beli {jumlah_beli}: {kemeja.hitung_total(jumlah_beli)}")
```

```python
# Parent Class
class Hewan:
    def __init__(self, nama, habitat):
        self.nama = nama
        self.habitat = habitat

    def info(self):
        print(f"Hewan: {self.nama}")
        print(f"Habitat: {self.habitat}")


# Child Class Kucing
class Kucing(Hewan):
    def __init__(self, nama, habitat, warna_bulu, ras):
        super().__init__(nama, habitat)
        self.warna_bulu = warna_bulu
        self.ras = ras

    def info(self):
        super().info()
        print(f"Warna Bulu: {self.warna_bulu}")
        print(f"Ras: {self.ras}")


# Child Class Paus
class Paus(Hewan):
    def __init__(self, nama, habitat, panjang, jenis):
        super().__init__(nama, habitat)
        self.panjang = panjang
        self.jenis = jenis

    def info(self):
        super().info()
        print(f"Panjang: {self.panjang} meter")
        print(f"Jenis: {self.jenis}")


# Child Class Singa
class Singa(Hewan):
    def __init__(self, nama, habitat, jumlah_kaki, suara):
        super().__init__(nama, habitat)
        self.jumlah_kaki = jumlah_kaki
        self.suara = suara

    def info(self):
        super().info()
        print(f"Jumlah Kaki: {self.jumlah_kaki}")
        print(f"Suara: {self.suara}")


# --- Running Program ---
kucing = Kucing("Kucing Persia", "Daratan", "Putih", "Persia")
paus = Paus("Paus Biru", "Laut", 25, "Biru")
singa = Singa("Singa Afrika", "Sabana", 4, "AUM")

print("=== Info Kucing ===")
kucing.info()
print("\n=== Info Paus ===")
paus.info()
print("\n=== Info Singa ===")
singa.info()
```

```python

class Produk:
    def __init__(self, nama, harga):
        self.nama = nama
        self.harga = harga

    def info(self):
        print(f"Produk: {self.nama}")
        print(f"Harga Asli: {self.harga}")


class Elektronik(Produk):
    def __init__(self, nama, harga, garansi):
        super().__init__(nama, harga)
        self.garansi = garansi

    def diskon(self):
        return self.harga * 0.9  # diskon 10%

    def info(self):
        super().info()
        print(f"Garansi: {self.garansi} bulan")
        print(f"Harga Setelah Diskon: {self.diskon()}")


class Pakaian(Produk):
    def __init__(self, nama, harga, ukuran):
        super().__init__(nama, harga)
        self.ukuran = ukuran

    def diskon(self):
        return self.harga * 0.8  # diskon 20%

    def info(self):
        super().info()
        print(f"Ukuran: {self.ukuran}")
        print(f"Harga Setelah Diskon: {self.diskon()}")


# --- Running Program ---
laptop = Elektronik("Laptop Gaming", 15000000, 24)
kemeja = Pakaian("Kemeja Flanel", 200000, "M")

print("=== Info Elektronik ===")
laptop.info()
print("\n=== Info Pakaian ===")
kemeja.info()
```
